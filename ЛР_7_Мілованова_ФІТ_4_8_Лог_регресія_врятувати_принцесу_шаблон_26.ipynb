На основному етапі **не видаляйте викиди автоматично**. За бажанням пізніше проведіть окремий експеримент із видаленням тільки з навчальної вибірки.

```python
# TODO: обчисліть q1, q3, iqr для X_train
# TODO: знайдіть кількість викидів за межами [q1 - 1.5*iqr, q3 + 1.5*iqr]
# TODO: відсортуйте результат за спаданням
```

```python
Q1 = X_train.quantile(0.25)
Q3 = X_train.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = ((X_train < lower_bound) | (X_train > upper_bound)).sum()

outliers_sorted = outliers.sort_values(ascending=False)
print("Кількість потенційних викидів у кожній ознаці:")
print(outliers_sorted)
```

Не варто без перевірки видаляти всі такі рядки з наступних причин:

Природа даних: Ці викиди можуть бути не помилками вимірювання, а реальними, хоч і рідкісними, екстремальними значеннями (наприклад, специфічними рецептами сильних зіллей).

Втрата корисного сигналу: Саме ці «аномальні» спостереження можуть містити ключову інформацію для правильного передбачення складних або нестандартних випадків (вилікується принцеса чи ні).

### 9. Логістична регресія

Навчіть `LogisticRegression` на стандартизованих навчальних даних. Оцініть модель на тестовій вибірці: **Accuracy**, `classification_report`.

```python
# TODO: створіть model = LogisticRegression()
# TODO: model.fit(...)
# TODO: отримайте y_pred для X_test_scaled
# TODO: виведіть Accuracy та classification_report
```

```python
model = LogisticRegression(random_state=42)
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

_Яких помилок більше? Чи достатньо однієї метрики Accuracy?_

Більше помилок другого роду (хибнонегативних) — модель частіше (у 26 випадках) помилково передбачає, що принцеса не вилікувана (0), хоча насправді вона вилікувана (1). Помилок першого роду (хибнопозитивних), коли модель каже "вилікувано" (1) для невилікуваної принцеси (0), трохи менше — 24 випадки.


Ні, однієї метрики Accuracy недостатньо. Хоча вона показує загальну частку правильних відповідей, вона не враховує ціну помилки. У задачі "Врятувати принцесу" хибнонегативна помилка (коли модель каже, що ліки не подіяли, а вони подіяли) може призвести до непотрібного пошуку нових інгредієнтів. А хибнопозитивна помилка (коли модель каже, що принцеса здорова, а вона насправді ні) може бути фатальною. Тому важливо аналізувати матрицю помилок та використовувати метрики Precision (щоб мінімізувати хибнопозитивні) та Recall (щоб мінімізувати хибнонегативні результати).
